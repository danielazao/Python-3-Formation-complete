```python
from datetime import datetime, date

# Demande de la date de naissance
naissance_str = input("Entrez votre date de naissance (format JJ/MM/AAAA) : ")
naissance = datetime.strptime(naissance_str, "%d/%m/%Y").date()

# Date du jour
aujourdhui = date.today()

# Calcul de la différence totale en jours
difference = aujourdhui - naissance
jours = difference.days

# Conversion
annees = jours // 365
mois = jours // 30
semaines = jours // 7

print(f"Vous avez environ :")
print(f"- {annees} ans")
print(f"- {mois} mois")
print(f"- {semaines} semaines")
print(f"- {jours} jours")
```

```python
prenom = input("Entre ton prénom : ")

# On met le prénom en minuscule pour faciliter le traitement
prenom_normalise = prenom.lower()

# Quelques surnoms possibles
surnoms = {
    "alex": "Lexou",
    "daniel": "Dany",
    "marie": "Mimi",
    "paul": "Polo",
    "sarah": "Sasa",
    "luc": "Lulu"
}

# Si le prénom existe dans le dictionnaire → surnom associé
if prenom_normalise in surnoms:
    surnom = surnoms[prenom_normalise]
else:
    # Sinon on crée un surnom automatiquement
    surnom = prenom.capitalize() + "ou"

print(f"Salut {prenom}, ton surnom sera : {surnom}")
```
