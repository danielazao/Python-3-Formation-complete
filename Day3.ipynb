```python
print("=== CALCULATRICE ===")
print("Choisissez une opération :")
print("1 - Addition")
print("2 - Soustraction")
print("3 - Multiplication")
print("4 - Division")

choix = input("Votre choix : ")

# On demande les nombres seulement après le choix
a = float(input("Entrez le premier nombre : "))
b = float(input("Entrez le deuxième nombre : "))

if choix == "1":
    resultat = a + b
    operation = "addition"

elif choix == "2":
    resultat = a - b
    operation = "soustraction"

elif choix == "3":
    resultat = a * b
    operation = "multiplication"

elif choix == "4":
    if b != 0:
        resultat = a / b
        operation = "division"
    else:
        print("Erreur : division par zéro impossible.")
        exit()

else:
    print("Choix invalide.")
    exit()

print(f"Le résultat de la {operation} est : {resultat}")
```
