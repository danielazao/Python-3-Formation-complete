```python
def comparer_deux_nombres():
    a = float(input("Entrez le premier nombre : "))
    b = float(input("Entrez le deuxième nombre : "))

    # Comparaison
    if a > b:
        print(f"{a} est plus grand que {b}")
    elif b > a:
        print(f"{b} est plus grand que {a}")
    else:
        print("Les deux nombres sont égaux")

    # Vérification signe
    for n in (a, b):
        if n > 0:
            print(f"{n} est positif")
        elif n < 0:
            print(f"{n} est négatif")
        else:
            print(f"{n} est égal à zéro")


def comparer_trois_nombres():
    a = float(input("Entrez le premier nombre : "))
    b = float(input("Entrez le deuxième nombre : "))
    c = float(input("Entrez le troisième nombre : "))

    maximum = max(a, b, c)
    minimum = min(a, b, c)

    print(f"Le plus grand nombre est : {maximum}")
    print(f"Le plus petit nombre est : {minimum}")


# Programme principal avec relance
while True:
    print("\n=== MENU COMPARAISON ===")
    print("1 - Comparer 2 nombres")
    print("2 - Comparer 3 nombres")
    print("3 - Quitter")

    choix = input("Votre choix : ")

    if choix == "1":
        comparer_deux_nombres()

    elif choix == "2":
        comparer_trois_nombres()

    elif choix == "3":
        print("Au revoir")
        break

    else:
        print("Choix invalide")

    # Option pour relancer
    relance = input("\nVoulez-vous recommencer ? (o/n) : ").lower()
    if relance != "o":
        print("Fin du programme")
        break
```
